# 반복문 - for

>단축키
- text 셀 -> code 셀로 바꾸기 : [esc -> y]
- code 셀 -> text 셀로 바꾸기 : [esc -> m]
- 실행 : ctrl + enter
- 실행 + 다음 셀로 커서 이동 : shift + enter
- 셀 삭제 : [esc -> dd]
- 삭제 되돌리기 : z
- 현재 셀의 위에 셀 추가 : a
- 햔재 셀의 아래에 셀 츠가 : b

```python
nums = [32, 65, 9, 72 ,800, 99, 1, 999, 199, 201]
print(len(nums))
```

```python
# 100 이상의 숫자만 출력
for num in nums:
    # print(num) -> 리스트 전체 순회하면서 출력
    if num >= 100:
        print(f"- 100 이상의 수: {num}")
```

```python
# 100 이상의 숫자, 100 미만인 숫자를 따로 새로운 리스트에 넣어서 각각 리스트로 출력
a = []
b = []
for num in nums:
    if num >= 100:
        a.append(num)
    else:
        b.append(num)
print(f"- 100 이상의 수: {a}")
print(f"- 100 미만의 수: {b}")
        
```

- for + dict

```python
# nums의 숫자가 각각 몇 번 반복하는지 dict로 만들어서 출력
nums = [1, 2, 6, 7, 4, 3, 2, 1, 5, 9, 4, 9, 7, 2, 1, 3, 4, 5, 6, 7, 5, 6, 7, 5]
```

```python
counter = {}

for num in nums:
    if num not in counter:
    # 특정 숫자가 제일 처음 나오면 그건 키에다가 값에 +1
        counter[num] = 1
    else:
    # 특정 숫자가 그 이후에 나오면 그건 값에 +1
        counter[num] = counter[num] + 1

print(counter)
```

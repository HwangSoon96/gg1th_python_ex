## 함수문제1

### 함수 정의


```python
# 매개변수로 전달된 값들을 모두 곱해서 리턴하는 가변 매개변수 함수를 만들어보세요.

def mul(*values):

    result = 1

    for i in values:
        result = result * i
    return result
```

### 호출부

```python
result = mul(5, 7, 9, 10)
print(result)
result = mul(5, 7, 9, 10, 11)
print(result)
result = mul(5, 7, 9, 10, 11, 12)
print(result)
result = mul(5, 7, 9, 10, 11, 16)
print(result)
```
